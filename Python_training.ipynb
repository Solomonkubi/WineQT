### **2. Data Cleaning**

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
# also you might want to impute them with fillna()
```

```python
df.isna().sum()
```

```python
df['Name']
```

```python
# Standardize Name (Remove spaces & fix capitalization)
df['Name'] = df['Name'].str.strip().str.title()
df['Name']
```

```python
df['Age'].dtype
```

```python
# Convert Age to numeric (Invalid values become NaN)
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Age'].dtype
```

```python
df['Age']
```

```python
df['Salary'].unique()
```

```python
 df['Salary'].dtype
```

```python
# Fix Salary
df['Salary'] = df['Salary'].replace({
    'Eighty Thousand': 80000, 'Sixty Thousand': 60000, '90K': 90000, '70,000': 70000})

df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
print(df['Salary'].dtype)
df['Salary'].head()
```

```python
df['Remarks']
```

```python
# Standardize Remarks
df['Remarks'] = df['Remarks'].str.lower().replace({'n/a': 'unknown', 'none': 'unknown', '': 'unknown'})
df['Remarks']
```

```python
df['Date Joined'].dtype
```

```python
# Convert Date Joined to datetime format
df['Date Joined'] = pd.to_datetime(df['Date Joined'], errors='coerce')
df.info()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
#  Print cleaned DataFrame
print(df.head())
```

```python
# Convert column to datetime format, keeping NaT values for invalid/missing dates
df['Date Joined'] = pd.to_datetime(df['Date Joined'], errors='coerce', dayfirst=False)

# Standardize format to YYYY-MM-DD (this keeps NaT values intact)
df['Date Joined'] = df['Date Joined'].dt.strftime('%Y-%m-%d')

print(df)
```

### **3. Data Cleaning: Strings**


```python
## Applying String Methods to a List of Strings

# Sample list of strings
strings = ["  Hello World  ", "Python Programming", "12345", "hello123", "   ", "UPPER lower"]
strings
```

```python
# strip() - Removes leading and trailing whitespace
print([s.strip() for s in strings])
```

```python
# lower() - Converts all characters to lowercase
print([s.lower() for s in strings])
```

```python
# upper() - Converts all characters to uppercase
print([s.upper() for s in strings])
```

```python
# replace() - Replaces occurrences of 'o' with '*'
print([s.replace("o", "*") for s in strings])
```

```python
# split() - Splits strings by space
print([s.split(" ") for s in strings])
```

```python
# join() - Joins a list of words into a single string
print(["-".join(s.split()) for s in strings])
```

```python
# find() - Finds the index of 'o'
print([s.find("o") for s in strings])
```

```python
# count() - Counts occurrences of 'o'
print([s.count("o") for s in strings])
```

```python
# startswith() - Checks if strings start with 'H'
print([s.startswith("H") for s in strings])
```

```python

# endswith() - Checks if strings end with 'd'
print([s.endswith("d") for s in strings])
```

```python

# format() - Formats a string
print(["This is: {}".format(s) for s in strings])
```

```python
# isalpha() - Checks if strings contain only letters
print([s.isalpha() for s in strings])
```

```python
# isdigit() - Checks if strings contain only digits
print([s.isdigit() for s in strings])
```

```python
# isalnum() - Checks if strings contain only alphanumeric characters
print([s.isalnum() for s in strings])
```

```python
# isspace() - Checks if strings contain only whitespace
print([s.isspace() for s in strings])
```

```python
# title() - Converts strings to title case
print([s.title() for s in strings])
```

```python
# capitalize() - Capitalizes first letter of string
print([s.capitalize() for s in strings])
```

```python
# swapcase() - Swaps upper/lower case characters
print([s.swapcase() for s in strings])
```

```python
# zfill(10) - Pads numeric strings with zeros
print([s.zfill(10) for s in strings])
```

```python
# rjust(15) - Right-justifies strings
print([s.rjust(15) for s in strings])
```

```python
# ljust(15) - Left-justifies strings
print([s.ljust(15) for s in strings])
```
